```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Step 1: Create the dataset
data = {
    'Size': [100, 200, 300, 400, 500],
    'Price': [50, 100, 150, 200, 250]
}

# Step 2: Create a DataFrame
df = pd.DataFrame(data)

print("House Price Dataset:")
print(df)

# Step 3: Define input (X) and output (y)
X = df[['Size']]
y = df['Price']

# Step 4: Create and train the model
model = LinearRegression()
model.fit(X, y)

# Step 5: Predict the price of a 350 sqm house
predicted_price = model.predict(pd.DataFrame({'Size': [350]}))

# Step 6: Display the prediction
print("\nPredicted price for a 350 sqm house:")
print(f"{predicted_price[0]} million")
```
